# Monty Hall Problem

The next code snippet demonstrates the counterintuitive Monty Hall problem using probability theory, exhaustive enumeration, and Monte Carlo simulation.

The problem:
1. Three doors conceal one car and two zonks. You select one door.
1. The host, who knows where the car is, opens another door revealing a zonk.
1. You may stay with your original choice or switch to the remaining closed door.

Mathematical result:

- P(*Win by Staying*) = 1 / 3
- P(*Win by Switching*) = 2 / 3

The crucial insight: Switching wins exactly when your initial choice was wrong.

```python
"""
Assumptions:
1. Exactly one of three doors contains a car.
2. The other two doors contain zonks.
3. The contestant initially chooses a door uniformly at random.
4. The host knows where the car is.
5. The host always opens an unchosen door containing a zonk.
6. The host always offers the contestant the opportunity to switch.
Question:
Should the contestant stay or switch?
"""

import random

# ============================================================
# 1. THEORETICAL ANALYSIS
# ============================================================
# There are three doors, each equally likely to contain the car.
#
# Probability that the initial choice is correct:
#
#   P(initially correct) = 1/3
#
# Probability that the initial choice is incorrect:
#
#   P(initially incorrect) = 2/3
#
# Staying wins precisely when the initial choice is correct.
# Switching wins precisely when the initial choice is incorrect.
#
# Therefore:
#
#   P(win | stay)   = 1/3
#   P(win | switch) = 2/3
print("THEORETICAL PROBABILITIES")
print("Stay:", 1 / 3)
print("Switch:", 2 / 3)

# ============================================================
# 2. EXHAUSTIVE ENUMERATION
# ============================================================
# We can establish the result by examining every possible
# location of the car while keeping the initial choice fixed.
#
# By symmetry, choosing door 0 loses no generality.
#
# Car location | Initial choice | Stay wins | Switch wins
# -------------------------------------------------------
#      0       |       0        |    Yes    |     No
#      1       |       0        |    No     |     Yes
#      2       |       0        |    No     |     Yes
#
# Switching wins in 2 out of 3 equally likely cases.
doors = [0, 1, 2]
initial_choice = 0
stay_wins = 0
switch_wins = 0
for car in doors:
    # Staying wins if the initial choice contains the car.
    if initial_choice == car:
        stay_wins += 1
    # Switching wins if the initial choice was incorrect.
    else:
        switch_wins += 1
print("\nEXHAUSTIVE ENUMERATION")
print("Stay wins:", stay_wins, "out of 3")
print("Switch wins:", switch_wins, "out of 3")

# ============================================================
# 3. SIMULATING ONE GAME
# ============================================================
def play_game():
    """
    Simulate one Monty Hall game.
    Returns:
        (stay_win, switch_win)
    Each value is a Boolean indicating whether that
    strategy wins.
    """
    doors = [0, 1, 2]
    # Randomly place the car behind one door.
    car = random.choice(doors)
    # Contestant randomly chooses an initial door.
    initial_choice = random.choice(doors)
    # The host must reveal a zonk behind an unchosen door.
    #
    # Importantly, the host NEVER reveals the car.
    # This assumption is essential to the problem.
    possible_reveals = [
        door
        for door in doors
        if door != initial_choice and door != car
    ]
    revealed_door = random.choice(possible_reveals)
    # Exactly two doors remain closed:
    # the initial choice and one other door.
    #
    # Switching means choosing the other closed door.
    remaining_doors = [
        door
        for door in doors
        if door != initial_choice and door != revealed_door
    ]
    switched_choice = remaining_doors[0]
    # Determine which strategies win.
    stay_win = (initial_choice == car)
    switch_win = (switched_choice == car)
    return stay_win, switch_win

# ============================================================
# 4. MONTE CARLO SIMULATION
# ============================================================
# Monte Carlo methods estimate probabilities using
# repeated random experiments.
#
# By the Law of Large Numbers:
#
#   observed frequency -> theoretical probability
#
# as the number of independent trials becomes large.
def simulate(n_trials):
    """Run n_trials independent Monty Hall games."""
    stay_wins = 0
    switch_wins = 0
    for _ in range(n_trials):
        stay_win, switch_win = play_game()
        stay_wins += stay_win
        switch_wins += switch_win
    # Estimate probabilities from relative frequencies.
    p_stay = stay_wins / n_trials
    p_switch = switch_wins / n_trials
    return p_stay, p_switch
# Run simulations with increasing numbers of trials.
trial_counts = [10, 100, 1000, 10000, 100000]
print("\nMONTE CARLO SIMULATION")
print(f"{'Trials':>10} {'Stay':>12} {'Switch':>12}")
print("-" * 36)
for n in trial_counts:
    p_stay, p_switch = simulate(n)
    print(f"{n:>10} {p_stay:>12.4f} {p_switch:>12.4f}")

# ============================================================
# 5. VERIFYING THE MATHEMATICAL RESULT
# ============================================================
# With sufficiently many trials, the simulated probabilities
# should approach 1/3 and 2/3.
N = 100000
p_stay, p_switch = simulate(N)
theoretical_stay = 1 / 3
theoretical_switch = 2 / 3
print("\nCOMPARISON WITH THEORY")
print("Stay:")
print("  Theoretical:", theoretical_stay)
print("  Simulated:  ", p_stay)
print("  Error:      ", abs(p_stay - theoretical_stay))
print("Switch:")
print("  Theoretical:", theoretical_switch)
print("  Simulated:  ", p_switch)
print("  Error:      ", abs(p_switch - theoretical_switch))

# ============================================================
# 6. WHY IS SWITCHING BETTER?
# ============================================================
# The initial choice has probability 1/3 of being correct.
#
# Therefore, the TWO unchosen doors together have
# probability 2/3 of containing the car.
#
# The host knows where the car is and deliberately reveals
# a zonk among those two doors.
#
# This transfers no probability to the initial choice.
#
# Instead, the remaining unchosen door retains the entire
# 2/3 probability under the stated host rules.
#
# The mistake is to assume:
#
#   "Two doors remain, so each must have probability 1/2."
#
# The host's action is NOT independent of the car's location.
#
# The two remaining doors are therefore NOT symmetric.

# ============================================================
# 7. GENERALIZATION: 100 DOORS
# ============================================================
# Imagine the game with 100 doors:
#
# - One door contains a car.
# - You initially choose one door.
# - The host reveals 98 zonks.
# - Exactly two closed doors remain.
#
# Initially:
#
#   P(initial choice correct) = 1/100
#
#   P(initial choice wrong)   = 99/100
#
# Therefore:
#
#   P(win by staying)   = 0.01
#
#   P(win by switching) = 0.99
#
# The advantage of switching becomes much more intuitive.
def simulate_many_doors(n_doors, n_trials):
    """
    Generalized Monty Hall simulation.
    The host opens n_doors - 2 zonk doors,
    leaving the initial choice and one other door closed.
    """
    switch_wins = 0
    for _ in range(n_trials):
        car = random.randrange(n_doors)
        choice = random.randrange(n_doors)
        # Under the host's rules, switching wins
        # exactly when the initial choice is incorrect.
        if choice != car:
            switch_wins += 1
    return switch_wins / n_trials
print("\nGENERALIZATION TO 100 DOORS")
p_switch_100 = simulate_many_doors(100, 100000)
print("Theoretical switch probability:", 99 / 100)
print("Simulated switch probability:  ", p_switch_100)

# ============================================================
# CONCLUSION
# ============================================================
# The Monty Hall problem illustrates three important ideas:
#
# 1. Conditional probability can contradict intuition.
#
# 2. Information revealed by a knowledgeable host
#    is not equivalent to a random event.
#
# 3. Computational experiments can verify and illustrate
#    mathematical results, but they do not replace proofs.
#
# FINAL RESULT:
#
#   ALWAYS SWITCH.
#
# Under the stated assumptions, switching doubles
# the probability of winning from 1/3 to 2/3.
```
